### Indexing & Slicing

```python
villes = ["Genève", "Lausanne", "Renens", "Yverdon"]

## Indexing
print(villes[0])    # premier élément, index 0
print(villes[-1])   # dernier élément

## Slicing idx[début:fin:pas]
print(villes[:2])     # début par défaut = 0, s'arrête avant l'index 2
print(villes[2:])     # commence à l'index 2, fin par défaut = jusqu'au bout
print(villes[:2:2])   # début par défaut = 0, s'arrête avant l'index 2, pas de 2 -> juste l'index 0
print(villes[::3])    # début par défaut = 0, fin par défaut = jusqu'au bout, pas de 3 -> index 0 et 3
print(villes[::-1])   # début et fin par défaut, pas de -1 -> parcourt à l'envers, du dernier au premier
```

### Méthodes de liste : append, insert, extend, remove, pop, sort, sorted, reverse, index, count, clear

```python
villes = ["Genève", "Lausanne", "Renens", "Yverdon"]

## append : ajoute un élément à la fin
villes.append("Nyon")
print(villes)   # ['Genève', 'Lausanne', 'Renens', 'Yverdon', 'Nyon']

## insert : ajoute un élément à une position précise
villes.insert(1, "Sion")
print(villes)   # ['Genève', 'Sion', 'Lausanne', 'Renens', 'Yverdon', 'Nyon']

## extend : ajoute tous les éléments d'un autre itérable (pas un seul objet)
villes.extend(["Bulle", "Fribourg"])
print(villes)   # [..., 'Nyon', 'Bulle', 'Fribourg']

## append vs extend sur une liste
test: list = ["a", "b"]
test.append(["c", "d"])
print(test)   # ['a', 'b', ['c', 'd']]  -> ajoute la liste ENTIÈRE comme un seul élément

test2 = ["a", "b"]
test2.extend(["c", "d"])
print(test2)   # ['a', 'b', 'c', 'd']  -> ajoute chaque élément séparément

## remove : enlève la première occurrence trouvée (par valeur, pas par index)
villes.remove("Sion")
print(villes)

## pop : enlève et retourne l'élément à un index donné (dernier par défaut)
dernier = villes.pop()
print(dernier, villes)

## sort : trie en place (modifie la liste directement, ne retourne rien)
villes.sort()
print(villes)   # ordre alphabétique

villes.sort(reverse=True)
print(villes)   # ordre alphabétique inversé

villes.sort(key=len)
print(villes)   # trié par longueur du nom de ville

## sorted : pareil que sort mais retourne une NOUVELLE liste, laisse l'originale intacte
nouvelle = sorted(villes)
print(nouvelle)
print(villes)   # inchangée

## reverse : inverse l'ordre en place
villes.reverse()
print(villes)

## index : retourne l'index de la première occurrence
print(villes.index("Genève"))

## count : compte les occurrences d'une valeur
print(villes.count("Genève"))

## clear : vide la liste
villes.clear()
print(villes)   # []
```

### `enumerate` : boucler avec l'index ET la valeur

Évite d'écrire `for i in range(len(liste)): liste[i]` à la main.

```python
villes = ["Genève", "Lausanne", "Renens", "Yverdon"]

for i, ville in enumerate(villes):
    print(i, ville)

## start : démarre le compteur à une autre valeur que 0
for i, ville in enumerate(villes, start=1):
    print(f"{i}. {ville}")
```

### `zip` : boucler sur plusieurs listes en parallèle

S'arrête à la longueur de la liste la plus courte.

```python
villes = ["Genève", "Lausanne", "Renens", "Yverdon"]
cantons = ["GE", "VD", "VD", "VD"]

for ville, canton in zip(villes, cantons):
    print(f"{ville} ({canton})")

## zip s'arrête sur la liste la plus courte
codes_postaux = [1200, 1000]
for ville, cp in zip(villes, codes_postaux):
    print(ville, cp)   # seulement 2 paires, Renens/Yverdon ignorés

## zip + enumerate combinés
for i, (ville, canton) in enumerate(zip(villes, cantons), start=1):
    print(f"{i}. {ville} - {canton}")

## dézipper : reconstruire les listes séparées à partir de paires
paires = list(zip(villes, cantons))
villes_only, cantons_only = zip(*paires)
print(list(villes_only))
print(list(cantons_only))
```

## Dictionnaires : accès, méthodes

Accès sûr avec `get`, fusion avec `update`/`|`, boucler avec `items()`.

```python
knight = {"nom": "Rey", "type": "Jedi", "force": 87}

## Accès
print(knight["nom"])          # accès direct, KeyError si la clé n'existe pas
print(knight.get("nom"))      # accès sûr, retourne None si absente (pas d'erreur)
print(knight.get("age", 0))   # valeur par défaut si absente

## Ajouter / modifier
knight["age"] = 25            # ajoute une clé si elle n'existe pas, sinon la modifie
print(knight)

## update : fusionne un autre dict (ou des kwargs) dedans
knight.update({"force": 92, "arme": "sabre laser"})
print(knight)

## keys, values, items
print(list(knight.keys()))
print(list(knight.values()))
print(list(knight.items()))   # liste de tuples (clé, valeur)

## Boucler sur un dict
for cle, valeur in knight.items():
    print(f"{cle}: {valeur}")

## pop : enlève une clé et retourne sa valeur
force = knight.pop("force")
print(force, knight)

## pop avec valeur par défaut : évite le KeyError si la clé n'existe pas
inexistant = knight.pop("clé_absente", "valeur_par_défaut")
print(inexistant)

## popitem : enlève et retourne le dernier couple (clé, valeur) inséré
dernier = knight.popitem()
print(dernier, knight)

## setdefault : retourne la valeur si la clé existe, sinon l'ajoute avec une valeur par défaut
knight.setdefault("planete", "Jakku")
print(knight)

## in : teste la présence d'une clé (pas d'une valeur)
print("nom" in knight)        # True
print("Rey" in knight)        # False, "Rey" est une valeur, pas une clé
print("Rey" in knight.values())  # True

## del : supprime une clé
del knight["age"]
print(knight)

## fromkeys : crée un dict à partir d'une liste de clés, valeur par défaut commune
noms = ["Rey", "Finn", "Poe"]
scores = dict.fromkeys(noms, 0)
print(scores)   # {'Rey': 0, 'Finn': 0, 'Poe': 0}

vide = dict.fromkeys(noms)
print(vide)   # {'Rey': None, 'Finn': None, 'Poe': None}

## piège fromkeys avec une valeur mutable : toutes les clés partagent la MÊME référence
pieges = dict.fromkeys(noms, [])
pieges["Rey"].append("sabre")
print(pieges)   # {'Rey': ['sabre'], 'Finn': ['sabre'], 'Poe': ['sabre']} -> les 3 changent

## Fusionner deux dicts (Python 3.9+)
d1 = {"a": 1, "b": 2}
d2 = {"b": 99, "c": 3}
fusion = d1 | d2
print(fusion)   # {'a': 1, 'b': 99, 'c': 3} -> d2 écrase les clés communes de d1

## clear
knight.clear()
print(knight)   # {}
```

### List comprehension

```python
## basique : [expression for élément in itérable]
carres = [n**2 for n in range(5)]
print(carres)   # [0, 1, 4, 9, 16]

## avec condition (filtre)
pairs = [n for n in range(10) if n % 2 == 0]
print(pairs)   # [0, 2, 4, 6, 8]

## if/else en expression (ternaire) : différent du filtre, garde tous les éléments
parite = ["pair" if n % 2 == 0 else "impair" for n in range(5)]
print(parite)   # ['pair', 'impair', 'pair', 'impair', 'pair']

## transformer une liste existante
villes = ["Genève", "Lausanne", "Renens"]
majuscules = [ville.upper() for ville in villes]
print(majuscules)   # ['GENÈVE', 'LAUSANNE', 'RENENS']

## comprehension imbriquée (équivalent à deux boucles for imbriquées)
grille = [(i, j) for i in range(3) for j in range(3)]
print(grille)   # [(0, 0), (0, 1), (0, 2), (1, 0), ...]

## avec zip : combiner deux listes
cantons = ["GE", "VD", "VD"]
combo = [f"{v} ({c})" for v, c in zip(villes, cantons)]
print(combo)   # ['Genève (GE)', 'Lausanne (VD)', 'Renens (VD)']
```

### Dict comprehension

```python
## basique : {clé: valeur for ... in ...}
carres = {n: n**2 for n in range(5)}
print(carres)   # {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}

## avec condition (filtre)
pairs_carres = {n: n**2 for n in range(10) if n % 2 == 0}
print(pairs_carres)   # {0: 0, 2: 4, 4: 16, 6: 36, 8: 64}

## valeur par défaut propre (alternative à fromkeys pour du mutable)
noms = ["Rey", "Finn", "Poe"]
inventaires = {nom: [] for nom in noms}
inventaires["Rey"].append("sabre")
print(inventaires)   # {'Rey': ['sabre'], 'Finn': [], 'Poe': []} -> indépendants

## transformer un dict existant (inverser clés/valeurs)
scores = {"Rey": 87, "Finn": 65, "Poe": 78}
inverse = {valeur: cle for cle, valeur in scores.items()}
print(inverse)   # {87: 'Rey', 65: 'Finn', 78: 'Poe'}

## depuis deux listes (avec zip)
villes = ["Genève", "Lausanne", "Renens"]
cantons = ["GE", "VD", "VD"]
villes_cantons = {ville: canton for ville, canton in zip(villes, cantons)}
print(villes_cantons)   # {'Genève': 'GE', 'Lausanne': 'VD', 'Renens': 'VD'}

## comprehension imbriquée
matrice = {(i, j): i * j for i in range(3) for j in range(3)}
print(matrice)   # {(0, 0): 0, (0, 1): 0, ..., (2, 2): 4}

## Exo Soluce
carre_paires = {str(k): k**2 for k in range(0, 20)}
print(pairs_carres)
```
