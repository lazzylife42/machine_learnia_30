### Indexing & Slicing

````python
villes = ["Genève", "Lausanne", "Renens", "Yverdon"]

## Indexing
print(villes[0])    # premier élément, index 0
print(villes[-1])   # dernier élément

## Slicing idx[début:fin:pas]
print(villes[:2])     # début par défaut = 0, s'arrête avant l'index 2
print(villes[2:])     # commence à l'index 2, fin par défaut = jusqu'au bout
print(villes[:2:2])   # début par défaut = 0, s'arrête avant l'index 2, pas de 2 -> juste l'index 0
print(villes[::3])    # début par défaut = 0, fin par défaut = jusqu'au bout, pas de 3 -> index 0 et 3
print(villes[::-1])   # début et fin par défaut, pas de -1 -> parcourt à l'envers, du dernier au premier
````

### Méthodes de liste : append, insert, extend, remove, pop, sort, sorted, reverse, index, count, clear

````python
villes = ["Genève", "Lausanne", "Renens", "Yverdon"]

## append : ajoute un élément à la fin
villes.append("Nyon")
print(villes)   # ['Genève', 'Lausanne', 'Renens', 'Yverdon', 'Nyon']

## insert : ajoute un élément à une position précise
villes.insert(1, "Sion")
print(villes)   # ['Genève', 'Sion', 'Lausanne', 'Renens', 'Yverdon', 'Nyon']

## extend : ajoute tous les éléments d'un autre itérable (pas un seul objet)
villes.extend(["Bulle", "Fribourg"])
print(villes)   # [..., 'Nyon', 'Bulle', 'Fribourg']

## append vs extend sur une liste
test: list = ["a", "b"]
test.append(["c", "d"])
print(test)   # ['a', 'b', ['c', 'd']]  -> ajoute la liste ENTIÈRE comme un seul élément

test2 = ["a", "b"]
test2.extend(["c", "d"])
print(test2)   # ['a', 'b', 'c', 'd']  -> ajoute chaque élément séparément

## remove : enlève la première occurrence trouvée (par valeur, pas par index)
villes.remove("Sion")
print(villes)

## pop : enlève et retourne l'élément à un index donné (dernier par défaut)
dernier = villes.pop()
print(dernier, villes)

## sort : trie en place (modifie la liste directement, ne retourne rien)
villes.sort()
print(villes)   # ordre alphabétique

villes.sort(reverse=True)
print(villes)   # ordre alphabétique inversé

villes.sort(key=len)
print(villes)   # trié par longueur du nom de ville

## sorted : pareil que sort mais retourne une NOUVELLE liste, laisse l'originale intacte
nouvelle = sorted(villes)
print(nouvelle)
print(villes)   # inchangée

## reverse : inverse l'ordre en place
villes.reverse()
print(villes)

## index : retourne l'index de la première occurrence
print(villes.index("Genève"))

## count : compte les occurrences d'une valeur
print(villes.count("Genève"))

## clear : vide la liste
villes.clear()
print(villes)   # []
````

### `enumerate` : boucler avec l'index ET la valeur

Évite d'écrire `for i in range(len(liste)): liste[i]` à la main.

````python
villes = ["Genève", "Lausanne", "Renens", "Yverdon"]

for i, ville in enumerate(villes):
    print(i, ville)

## start : démarre le compteur à une autre valeur que 0
for i, ville in enumerate(villes, start=1):
    print(f"{i}. {ville}")
````

### `zip` : boucler sur plusieurs listes en parallèle

S'arrête à la longueur de la liste la plus courte.

````python
villes = ["Genève", "Lausanne", "Renens", "Yverdon"]
cantons = ["GE", "VD", "VD", "VD"]

for ville, canton in zip(villes, cantons):
    print(f"{ville} ({canton})")

## zip s'arrête sur la liste la plus courte
codes_postaux = [1200, 1000]
for ville, cp in zip(villes, codes_postaux):
    print(ville, cp)   # seulement 2 paires, Renens/Yverdon ignorés

## zip + enumerate combinés
for i, (ville, canton) in enumerate(zip(villes, cantons), start=1):
    print(f"{i}. {ville} - {canton}")

## dézipper : reconstruire les listes séparées à partir de paires
paires = list(zip(villes, cantons))
villes_only, cantons_only = zip(*paires)
print(list(villes_only))
print(list(cantons_only))
````

## Dictionnaires : accès, méthodes

Accès sûr avec `get`, fusion avec `update`/`|`, boucler avec `items()`.

````python
knight = {"nom": "Rey", "type": "Jedi", "force": 87}

## Accès
print(knight["nom"])          # accès direct, KeyError si la clé n'existe pas
print(knight.get("nom"))      # accès sûr, retourne None si absente (pas d'erreur)
print(knight.get("age", 0))   # valeur par défaut si absente

## Ajouter / modifier
knight["age"] = 25            # ajoute une clé si elle n'existe pas, sinon la modifie
print(knight)

## update : fusionne un autre dict (ou des kwargs) dedans
knight.update({"force": 92, "arme": "sabre laser"})
print(knight)

## keys, values, items
print(list(knight.keys()))
print(list(knight.values()))
print(list(knight.items()))   # liste de tuples (clé, valeur)

## Boucler sur un dict
for cle, valeur in knight.items():
    print(f"{cle}: {valeur}")

## pop : enlève une clé et retourne sa valeur
force = knight.pop("force")
print(force, knight)

## pop avec valeur par défaut : évite le KeyError si la clé n'existe pas
inexistant = knight.pop("clé_absente", "valeur_par_défaut")
print(inexistant)

## popitem : enlève et retourne le dernier couple (clé, valeur) inséré
dernier = knight.popitem()
print(dernier, knight)

## setdefault : retourne la valeur si la clé existe, sinon l'ajoute avec une valeur par défaut
knight.setdefault("planete", "Jakku")
print(knight)

## in : teste la présence d'une clé (pas d'une valeur)
print("nom" in knight)        # True
print("Rey" in knight)        # False, "Rey" est une valeur, pas une clé
print("Rey" in knight.values())  # True

## del : supprime une clé
del knight["age"]
print(knight)

## fromkeys : crée un dict à partir d'une liste de clés, valeur par défaut commune
noms = ["Rey", "Finn", "Poe"]
scores = dict.fromkeys(noms, 0)
print(scores)   # {'Rey': 0, 'Finn': 0, 'Poe': 0}

vide = dict.fromkeys(noms)
print(vide)   # {'Rey': None, 'Finn': None, 'Poe': None}

## piège fromkeys avec une valeur mutable : toutes les clés partagent la MÊME référence
pieges = dict.fromkeys(noms, [])
pieges["Rey"].append("sabre")
print(pieges)   # {'Rey': ['sabre'], 'Finn': ['sabre'], 'Poe': ['sabre']} -> les 3 changent

## Fusionner deux dicts (Python 3.9+)
d1 = {"a": 1, "b": 2}
d2 = {"b": 99, "c": 3}
fusion = d1 | d2
print(fusion)   # {'a': 1, 'b': 99, 'c': 3} -> d2 écrase les clés communes de d1

## clear
knight.clear()
print(knight)   # {}
````

### List comprehension

````python
## basique : [expression for élément in itérable]
carres = [n**2 for n in range(5)]
print(carres)   # [0, 1, 4, 9, 16]

## avec condition (filtre)
pairs = [n for n in range(10) if n % 2 == 0]
print(pairs)   # [0, 2, 4, 6, 8]

## if/else en expression (ternaire) : différent du filtre, garde tous les éléments
parite = ["pair" if n % 2 == 0 else "impair" for n in range(5)]
print(parite)   # ['pair', 'impair', 'pair', 'impair', 'pair']

## transformer une liste existante
villes = ["Genève", "Lausanne", "Renens"]
majuscules = [ville.upper() for ville in villes]
print(majuscules)   # ['GENÈVE', 'LAUSANNE', 'RENENS']

## comprehension imbriquée (équivalent à deux boucles for imbriquées)
grille = [(i, j) for i in range(3) for j in range(3)]
print(grille)   # [(0, 0), (0, 1), (0, 2), (1, 0), ...]

## avec zip : combiner deux listes
cantons = ["GE", "VD", "VD"]
combo = [f"{v} ({c})" for v, c in zip(villes, cantons)]
print(combo)   # ['Genève (GE)', 'Lausanne (VD)', 'Renens (VD)']
````

### Dict comprehension

````python
## basique : {clé: valeur for ... in ...}
carres = {n: n**2 for n in range(5)}
print(carres)   # {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}

## avec condition (filtre)
pairs_carres = {n: n**2 for n in range(10) if n % 2 == 0}
print(pairs_carres)   # {0: 0, 2: 4, 4: 16, 6: 36, 8: 64}

## valeur par défaut propre (alternative à fromkeys pour du mutable)
noms = ["Rey", "Finn", "Poe"]
inventaires = {nom: [] for nom in noms}
inventaires["Rey"].append("sabre")
print(inventaires)   # {'Rey': ['sabre'], 'Finn': [], 'Poe': []} -> indépendants

## transformer un dict existant (inverser clés/valeurs)
scores = {"Rey": 87, "Finn": 65, "Poe": 78}
inverse = {valeur: cle for cle, valeur in scores.items()}
print(inverse)   # {87: 'Rey', 65: 'Finn', 78: 'Poe'}

## depuis deux listes (avec zip)
villes = ["Genève", "Lausanne", "Renens"]
cantons = ["GE", "VD", "VD"]
villes_cantons = {ville: canton for ville, canton in zip(villes, cantons)}
print(villes_cantons)   # {'Genève': 'GE', 'Lausanne': 'VD', 'Renens': 'VD'}

## comprehension imbriquée
matrice = {(i, j): i * j for i in range(3) for j in range(3)}
print(matrice)   # {(0, 0): 0, (0, 1): 0, ..., (2, 2): 4}
````

### Exécuter une commande shell depuis Python

`os` a plusieurs façons de lancer une commande, à ne pas confondre :
- `os.system(cmd)` : bloquant, retourne juste le code de sortie, pas la sortie texte
- `os.popen(cmd)` : donne accès à la sortie, mais déprécié
- `os.exec*()` (`execv`, `execvp`, ...) : REMPLACE le processus Python actuel, le script ne continue jamais après

Le standard moderne : `subprocess.run()`, plus sûr (pas d'injection shell si commande en liste) et donne stdout/stderr/returncode séparément.

```python
import subprocess

result = subprocess.run(["ls", "-la"], capture_output=True, text=True)
print(result.stdout)
print(result.returncode)
```

### Piège : `sep=` avec un seul argument
`sep` sépare plusieurs arguments passés à `print`, pas les éléments internes d'un seul objet (liste/dict). Sur un seul argument, `sep` ne fait rien.

````python
## Exo 8

import os
import glob

files = glob.glob("*")
## `glob.glob("*")` matche aussi les dossiers -> open() plante avec IsADirectoryError
files = [f for f in glob.glob("*") if os.path.isfile(f)]

## open(file, "r") plante en UnicodeDecodeError sur un fichier binaire (image, .pyc, etc.)
## -> mode "rb" si contenu potentiellement non-texte

backup = {}
for file in files:
    with open(file, "r") as f:
        backup[file] = f.read().splitlines()

for nom, contenu in backup.items():
    print(f"{nom}:\n{contenu}\n")
````

## PYTHON PROGRAMMATION ORIENTÉE OBJET (9/30)

Une classe = le plan de conception d'un objet. Un objet a des attributs (état) et des méthodes (actions qui peuvent modifier son état ou celui d'autres objets). Pas indispensable de créer ses propres classes en ML/data science, mais essentiel de comprendre le concept pour naviguer dans la doc et comprendre les objets qu'on manipule (`model.fit()`, `model.coef_`, etc. sont tous des attributs/méthodes d'objets).

````python
## --- Classe basique : attributs et méthodes ---
class Stylo:
    def __init__(self, couleur, marque):   # __init__ = constructeur, appelé à la création
        self.couleur = couleur              # attribut d'instance
        self.marque = marque
        self.encre_restante = 100           # attribut avec valeur par défaut

    def ecrire(self):                       # méthode = fonction qui appartient à l'objet
        if self.encre_restante > 0:
            self.encre_restante -= 10       # une méthode peut modifier l'état de l'objet
            print(f"Le stylo {self.couleur} écrit. Encre restante: {self.encre_restante}%")
        else:
            print("Plus d'encre !")


## instanciation : créer un objet à partir de la classe (le "plan")
stylo1 = Stylo("bleu", "Bic")
stylo2 = Stylo("rouge", "Pilot")

print(stylo1.couleur)   # accès à un attribut
stylo1.ecrire()          # appel d'une méthode
stylo1.ecrire()

print(stylo2.encre_restante)   # stylo2 est indépendant de stylo1, chaque instance a son propre état


## --- self : référence à l'instance elle-même ---
## self.attribut = attribut propre à CETTE instance précise, pas partagé entre objets
## c'est implicite : stylo1.ecrire() équivaut en interne à Stylo.ecrire(stylo1)


## --- Attribut de classe vs attribut d'instance ---
class Stylo2:
    unite = "%"   # attribut de CLASSE : partagé par TOUTES les instances

    def __init__(self, couleur):
        self.couleur = couleur   # attribut d'INSTANCE : propre à chaque objet

print(Stylo2.unite)   # accessible sans instancier
s = Stylo2("noir")
print(s.unite)         # accessible aussi depuis l'instance


## --- Héritage : une classe qui hérite d'une autre ---
class StyloGomme(Stylo):        # hérite de tous les attributs/méthodes de Stylo
    def effacer(self):           # méthode supplémentaire, propre à StyloGomme
        print(f"Le stylo {self.couleur} efface.")

s3 = StyloGomme("vert", "Frixion")
s3.ecrire()      # méthode héritée de Stylo
s3.effacer()     # méthode propre à StyloGomme


## --- __repr__ / __str__ : contrôler l'affichage d'un objet avec print() ---
class Stylo3:
    def __init__(self, couleur):
        self.couleur = couleur

    def __repr__(self):
        return f"Stylo({self.couleur})"

print(Stylo3("bleu"))   # Stylo(bleu) au lieu de <__main__.Stylo3 object at 0x...>

## --- numpy.ndarray : un objet avec attributs et méthodes, comme Stylo ---
import numpy as np

arr = np.array([[1, 2, 3], [4, 5, 6]])   # arr est une INSTANCE de la classe ndarray

## attributs : décrivent l'état de l'objet, pas des méthodes (pas de parenthèses)
print(arr.shape)     # (2, 3) -> dimensions
print(arr.dtype)     # int64 -> type des éléments
print(arr.ndim)      # 2 -> nombre de dimensions
print(arr.size)      # 6 -> nombre total d'éléments

## méthodes : actions, avec parenthèses, peuvent modifier ou retourner un nouvel objet
print(arr.sum())         # méthode -> retourne un scalaire
print(arr.mean())
print(arr.reshape(3, 2)) # méthode -> retourne un NOUVEL ndarray, ne modifie pas arr en place
print(arr.shape)          # arr inchangé, toujours (2, 3)

arr.sort()                # certaines méthodes modifient l'objet EN PLACE (comme list.sort())
print(arr)
````

## PYTHON NUMPY MACHINE LEARNING (10/30)

Le `ndarray` (n dimensions), l'objet central de numpy en ML/data science. Constructeurs (`np.array`, `np.zeros`, `np.ones`, `np.full`, `np.random.randn`, `np.linspace`, `np.arange`), attributs clés (`shape`, `size`, `dtype`), et méthodes de manipulation de forme (`vstack`, `hstack`, `concatenate`, `reshape`, `squeeze`, `ravel`).

````python
import numpy as np

## --- Le tableau numpy, dimensions et shape ---
arr1d = np.array([1, 2, 3])
arr2d = np.array([[1, 2, 3], [4, 5, 6]])
arr3d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print(arr1d.shape)   # (3,) -> 1 dimension, 3 éléments
print(arr2d.shape)   # (2, 3) -> 2 lignes, 3 colonnes
print(arr3d.shape)   # (2, 2, 2) -> 3 dimensions

print(arr2d.ndim)    # nombre de dimensions
print(arr2d.size)    # nombre total d'éléments (2*3 = 6)


## --- Constructeurs ---
print(np.zeros((2, 3)))          # tableau rempli de 0
print(np.ones((2, 3)))           # tableau rempli de 1
print(np.full((2, 3), 7))        # tableau rempli d'une valeur custom

## random.randn : distribution normale (loi normale centrée réduite, moyenne 0, std 1)
print(np.random.randn(2, 3))

## linspace : n valeurs régulièrement espacées entre début et fin (bornes incluses)
print(np.linspace(0, 10, 5))     # [0., 2.5, 5., 7.5, 10.]

## arange : comme range() mais retourne un ndarray, accepte les floats
print(np.arange(0, 10, 2))       # [0, 2, 4, 6, 8]
print(np.arange(0, 1, 0.25))     # [0., 0.25, 0.5, 0.75]


## --- dtype : définir le type des valeurs ---
## important pour la mémoire/perf : un int16 prend 4x moins de place qu'un int64
arr_int = np.array([1, 2, 3], dtype=np.int16)
arr_float = np.array([1, 2, 3], dtype=np.float64)
print(arr_int.dtype)
print(arr_float.dtype)

## conversion après coup
arr_converti = arr_int.astype(np.float32)
print(arr_converti.dtype)


## --- Assembler des tableaux ---
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print(np.vstack((a, b)))                 # empile verticalement (ajoute des lignes)
print(np.hstack((a, b)))                 # empile horizontalement (ajoute des colonnes)
print(np.concatenate((a, b), axis=0))    # équivalent vstack (axis=0 -> lignes)
print(np.concatenate((a, b), axis=1))    # équivalent hstack (axis=1 -> colonnes)


## --- Manipuler la forme ---
arr = np.arange(12)              # [0, 1, ..., 11], shape (12,)

print(arr.reshape(3, 4))         # réorganise en 3x4, même données, nouvelle shape
print(arr.reshape(2, 2, 3))      # réorganise en 3D

## squeeze : enlève les dimensions de taille 1 (utile après un reshape/opération qui en laisse)
arr_extra = np.array([[[1, 2, 3]]])   # shape (1, 1, 3)
print(arr_extra.squeeze().shape)      # (3,) -> dimensions inutiles supprimées

## ravel : aplatit un tableau n-D en 1D (retourne une vue, pas une copie -> attention si tu modifies)
arr2d = np.array([[1, 2], [3, 4]])
print(arr2d.ravel())              # [1, 2, 3, 4]

## --- Notation mathématique (rappel) ---
## Matrice A de dimensions m x n : m = lignes, n = colonnes
## Élément A[i, j] : i = index de ligne, j = index de colonne
## Stockage : row-major / "C order" (numpy par défaut) vs column-major / "Fortran order"


## --- shape est un tuple : accessible par index ---
arr = np.zeros((4, 6))
print(arr.shape)         # (4, 6)
print(arr.shape[0])      # 4 -> nombre de lignes
print(arr.shape[1])      # 6 -> nombre de colonnes

n_lignes, n_colonnes = arr.shape   # unpacking direct, pratique
print(n_lignes, n_colonnes)


## --- np.random.seed() : fixe la graine aléatoire, reproductibilité ---
np.random.seed(42)
print(np.random.randn(3))   # toujours les mêmes valeurs à chaque exécution avec seed=42

## randn : tirage suivant une loi normale centrée réduite (moyenne 0, std 1)
print(np.random.randn(2, 3))


## --- np.eye() : matrice identité ---
print(np.eye(3))
# [[1., 0., 0.],
#  [0., 1., 0.],
#  [0., 0., 1.]]

## eye accepte aussi une forme non carrée
print(np.eye(3, 5))   # 3 lignes, 5 colonnes, diagonale de 1 partant du coin haut-gauche
````
